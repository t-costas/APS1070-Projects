````python
### YOUR CODE HERE ###
#Step 1: Compute covariance matrix of dataframe
df_cov = np.cov(data_std.T)
print(df_cov.shape)
````

````python
#Step 2: Write function to get covariance matrix of the data frame and returns sorted eigenvalues and vectors
#Using code from APS1070_Tutorial_PCA.ipynb
def get_sorted_eigen(df_cov):
  #eigen decomposition
  eigenvalues, eigenvectors = np.linalg.eigh(df_cov)
  #sort in decreasing order of eigenvalues (largest first)
  args = (-eigenvalues).argsort()
  eigenvalues = eigenvalues[args]
  eigenvectors = eigenvectors[:, args]
  return eigenvalues, eigenvectors
````

````python
eigenvals, eigenvecs = get_sorted_eigen(df_cov)
print("Eigenvalues: ", eigenvals)
print("Eigenvectors: ", eigenvecs)
````

````python
#Step 3: Scree plot
#calculate explained variance - code from PCA tutorial
eigValSum = sum(eigenvals)
expVar = [eigV/eigValSum*100 for eigV in eigenvals] #how important is it
cumExpVar = np.cumsum(expVar)
print(expVar)
cumExpVar

plt.figure(figsize=(10, 6))
plt.bar(range(len(expVar)), expVar, label = "Explained Variance")
plt.plot(cumExpVar, 'r-o', label = "Cumulative Explained Variance")
plt.legend
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance (%)')
plt.title('Scree Plot of Principal Components')
plt.show()
````

````python
#Step 4: Calculate how many PCs are needed to cover 99.9% of dataset
target_var = 99.9
cum_var = 0
num = 0
#iterate through the explained variance until cumulative variance equals or is greater than 99.9
for var in expVar:
  cum_var += var
  num += 1
  if cum_var >= target_var:
    break
print("Number of PCs needed to cover 99.9% of dataset: ", num)
val = sum(expVar[:num])
print(val)
````

````python
#Step 5: Plot the first 16 principal components (Eigenvectors) as a time series (16 subplots, on the x-axis you have dates and on the y-axis you have the value of the PC element) .
def plotPCs(eigenvecs, nplots):
  fig = plt.figure(figsize=(15,4*nplots))
  gs = gridspec.GridSpec(nplots,1)
  for i in range(nplots):
      ax = plt.subplot(gs[i])
      ax.set_xlabel('Year')
      ax.set_ylabel('PC Value')

      #plot the Year vs GDP earned
      ax.plot(data_std.columns, eigenvecs[:,i])

      #rotate x-axis labels and adjust the layout
      ax.set_xticks(data_std.columns) #ticks are all years
      ax.set_xticklabels(data_std.columns, rotation=45, ha='right')
      ax.set_title('Value of Principal Component ' + str(i+1) + ' Over Time')
  plt.tight_layout()
  gs.update(top=0.95, hspace=0.5)
  plt.show()
plotPCs(eigenvecs, 16)
````

````python
def plot_country_figures(original_df, country_name):
  ### YOUR CODE HERE ###
  #get standardized dataset for reconstruction
  scaler = StandardScaler()
  standardized_df = pd.DataFrame(scaler.fit_transform(original_df), columns=original_df.columns, index=original_df.index)
  #Step 1: Plot the original time-series for the specified country
  plotTimeSeries(original_df, [country_name], standardized=False)

  #get covariance matrix
  df_cov = np.cov(standardized_df.T)

  #Step 2: Plot the incremental reconstruction of the original (not standardized) time-series for the specified country in a single plot
  eigenvals, eigenvecs = get_sorted_eigen(df_cov)

  plt.figure(figsize=(15, 5))
  #use projection matrix and W to calculate the reconstructions
  reconstructions = pd.DataFrame(columns = standardized_df.columns)
  #save sum of squared residuals
  ssr = []
  best_reconstruction = None
  best_pcs = 0
  min_ssr = float('inf')
  #add extra curve to plot the reconstruction that covers 99.9% of the data (5 components)
  for i in [1, 2, 4, 5, 8, 16]:
    #select up to the ith principal component for reconstruction
    W = eigenvecs[:, :i]
    projX = np.dot(standardized_df, W)
    reconstructed_std = np.dot(projX, W.T)
    reconstructed = scaler.inverse_transform(reconstructed_std)

    #store reconstruction for plotting
    country_index = original_df.index.get_loc(country_name)
    new_row = pd.Series(reconstructed[original_df.index.get_loc(country_name)], index=original_df.columns, name=f'PC1 to PC{i}')
    reconstructions = pd.concat([reconstructions, new_row.to_frame().T], ignore_index=False)

    #calculate sum of squared residual for this reconstruction
    ssr_value = np.sum((original_df.loc[country_name] - reconstructed[country_index])**2)
    ssr.append(ssr_value)
    if ssr_value < min_ssr:
      min_ssr = ssr_value
      best_reconstruction = reconstructed[country_index]
      best_pcs = i
    #plot the reconstructed data for specified country
    plt.plot(original_df.columns, reconstructed[country_index], label=f'PC1 to PC{i}')
  #add original curve
  plt.plot(original_df.columns, original_df.loc[country_name], color='black', linewidth='4', linestyle='dotted', label='Original')
  plt.xlabel('Year')
  plt.ylabel('PC Value')
  plt.xticks(standardized_df.columns, rotation=45, ha='right')
  plt.title('Incremental Reconstruction of ' + country_name + ' Time Series')
  plt.legend()
  plt.show()
  #Step 3: Plot the residual error for your best reconstruction with respect to the original time-series
  #from piazza: Finding the Sum of Squared Residuals for each reconstruction and selecting the reconstruction with the least SSR

  residuals = original_df.loc[country_name] - best_reconstruction
  plt.figure(figsize=(15, 5))
  plt.plot(residuals.index, residuals.values)
  plt.xlabel('Year')
  plt.ylabel('Residual Error')
  plt.xticks(standardized_df.columns, rotation=45, ha='right')
  plt.title('Residual Error of ' + country_name + ' Time Series for Best Reconstruction: ' + str(best_pcs) + ' PCs')
  plt.show()
  #Step 4: RMSE of the reconstruction as a function of the number of included components
  rmse = []
  num_components = np.arange(1, 11)
  #print(num_components)
  for i in num_components:
    W = eigenvecs[:, :i]
    projX = np.dot(standardized_df, W)
    reconstructed_std = np.dot(projX, W.T)
    reconstructed = scaler.inverse_transform(reconstructed_std)
    country_index = original_df.index.get_loc(country_name)
    rmse.append(np.sqrt(np.mean((original_df.loc[country_name] - reconstructed[country_index])**2)))
    #plot RMSE as a function of number of components
  plt.figure(figsize=(15, 5))
  plt.plot(num_components, rmse)
  plt.xlabel('Number of Components')
  plt.ylabel('RMSE')
  plt.title('RMSE of Reconstruction vs Number of Components - '+ country_name)
  plt.show()
````

````python
### YOUR CODE HERE ###
def plot_country_figures_SVD(original_df, country_name):
  ### YOUR CODE HERE ###
  #get standardized dataset for reconstruction
  scaler = StandardScaler()
  standardized_df = pd.DataFrame(scaler.fit_transform(original_df), columns=original_df.columns, index=original_df.index)
  #Step 1: Plot the original time-series for the specified country
  plotTimeSeries(original_df, [country_name], standardized=False)

  #Step 2: Plot the incremental reconstruction of the original (not standardized) time-series for the specified country in a single plot using SVD
  #compare standardized and non-standardized errors to see if there is a significant difference when reconstructed
  U, S, V = np.linalg.svd(original_df, full_matrices=False) #since we are only using up to 16 singular values, not necessary to create full matrices
  print(U.shape, S.shape, V.shape)
  U_std, S_std, V_std = np.linalg.svd(standardized_df, full_matrices =False)

  plt.figure(figsize=(15, 10))

  #save sum of squared residuals
  ssr = []
  ssr_std = []
  best_reconstruction = None
  best_reconstruction_std = None
  best_sv = 0
  best_sv_std = 0
  min_ssr = float('inf')
  min_ssr_std = float('inf')

  reconstructions = pd.DataFrame(columns = standardized_df.columns)
  reconstructions_std = pd.DataFrame(columns = standardized_df.columns)
  #add extra curve to plot the reconstruction that covers 99.9% of the data (5 components)
  for i in [1, 2, 4, 5, 8, 16]:
    #select up to the ith U, S, and V component for reconstruction
    #referencing method from APS1070_Week_9_Lecture_Code.ipynb -- Toy Data reconstruction but expanding to include more components
    reconstructed = U[:, :i] @ np.diag(S[:i]) @ V[:i, :]
    reconstructed_std = U_std[:, :i] @ np.diag(S_std[:i]) @ V_std[:i, :]

    #scale reconstructed_std back to non-standardized
    reconstructed_std = scaler.inverse_transform(reconstructed_std)

    #save reconstructed PCs to dataframe series for plotting
    country_index = original_df.index.get_loc(country_name)
    new_row = pd.Series(reconstructed[original_df.index.get_loc(country_name)], index=original_df.columns, name=f'{i} Singular Value(s)')
    new_row_std = pd.Series(reconstructed_std[original_df.index.get_loc(country_name)], index=original_df.columns, name=f'{i} Singular Value(s)')
    reconstructions = pd.concat([reconstructions, new_row.to_frame().T], ignore_index=False)
    reconstructions_std = pd.concat([reconstructions_std, new_row_std.to_frame().T], ignore_index=False)

    #calculate sum of squared residual for this reconstruction
    ssr_value = np.sum((original_df.loc[country_name] - reconstructed[country_index])**2)
    ssr.append(ssr_value)
    if ssr_value < min_ssr:
      min_ssr = ssr_value
      best_reconstruction = reconstructed[country_index]
      best_sv = i
    ssr_value_std = np.sum((original_df.loc[country_name] - reconstructed_std[country_index])**2)
    ssr_std.append(ssr_value_std)

    if ssr_value_std < min_ssr_std:
      min_ssr_std = ssr_value_std
      best_reconstruction_std = reconstructed_std[country_index]
      best_sv_std = i

    #plot the reconstructed data for specified country
    plt.plot(original_df.columns, reconstructed[country_index], label=f'{i} Singular Value(s)')
    plt.plot(original_df.columns, reconstructed_std[country_index], linestyle='dashed', label=f'{i} Singular Value(s) - Standardized')
  #add original curve
  plt.plot(original_df.columns, original_df.loc[country_name], color='black', linewidth='4', linestyle='dotted', label='Original')
  plt.xlabel('Year')
  plt.ylabel('PC Value')
  plt.xticks(standardized_df.columns, rotation=45, ha='right')
  plt.title('Incremental Reconstruction of ' + country_name + ' Time Series')
  plt.legend()
  plt.show()
  #Step 3: Plot the residual error for your best reconstruction with respect to the original time-series
  #from piazza: Finding the Sum of Squared Residuals for each reconstruction and selecting the reconstruction with the least SSR

  residuals = original_df.loc[country_name] - best_reconstruction
  residuals_std = original_df.loc[country_name] - best_reconstruction_std
  #print(residuals)
  plt.figure(figsize=(15, 5))
  plt.plot(residuals.index, residuals.values)
  plt.plot(residuals_std.index, residuals_std.values, linestyle='dashed')
  plt.xlabel('Year')
  plt.ylabel('Residual Error')
  plt.legend(['Non-Standardized', 'Standardized'])
  plt.xticks(standardized_df.columns, rotation=45, ha='right')
  if best_sv == best_sv_std:
    plt.title('Residual Error of ' + country_name + ' Time Series for Best Reconstruction ' + str(best_sv) + ' Singular Values')
  else:
    plt.title('Residual Error of ' + country_name + ' Time Series for Best Reconstruction ' + str(best_sv) + '/' + str(best_sv_std) +' Singular Values')
  plt.show()
  #Step 4: RMSE of the reconstruction as a function of the number of included components
  rmse = []
  rmse_std = []
  num_components = np.arange(1, 11)
  #print(num_components)
  for i in num_components:
    reconstructed = U[:, :i] @ np.diag(S[:i]) @ V[:i, :]
    reconstructed_std = U_std[:, :i] @ np.diag(S_std[:i]) @ V_std[:i, :]
    reconstructed_std = scaler.inverse_transform(reconstructed_std)
    country_index = original_df.index.get_loc(country_name)
    rmse.append(np.sqrt(np.mean((original_df.loc[country_name] - reconstructed[country_index])**2)))
    rmse_std.append(np.sqrt(np.mean((original_df.loc[country_name] - reconstructed_std[country_index])**2)))
    #plot RMSE as a function of number of components
  plt.figure(figsize=(15, 5))
  plt.plot(num_components, rmse)
  plt.plot(num_components, rmse_std, linestyle='dashed')
  plt.legend(['Non-Standardized', 'Standardized'])
  plt.xlabel('Number of Components')
  plt.ylabel('RMSE')
  plt.title('RMSE of Reconstruction vs Number of Components - '+ country_name)
  plt.show()
````

````python
#Function to compare PCA and SVD results
def compare_PCA_SVD(original_df, country_name):
  #get standardized dataset for reconstruction
  scaler = StandardScaler()
  standardized_df = pd.DataFrame(scaler.fit_transform(original_df), columns=original_df.columns, index=original_df.index)
  #compute covariance matrix
  df_cov = np.cov(standardized_df.T)
  #PCA Eigenvectors/values
  eigenvals, eigenvecs = get_sorted_eigen(df_cov)
  #SVD matrices
  U, S, V = np.linalg.svd(original_df, full_matrices=False)
  #SVD matrices - standardized
  U_std, S_std, V_std = np.linalg.svd(standardized_df, full_matrices=False)

  #reconstruction on best residual - 16 components
  comp = 16
  plt.figure(figsize=(15, 5))

  reconstructed = U[:, :comp] @ np.diag(S[:comp]) @ V[:comp, :]
  reconstructed_std = U_std[:, :comp] @ np.diag(S_std[:comp]) @ V_std[:comp, :]
  reconstructed_std = scaler.inverse_transform(reconstructed_std)

  W = eigenvecs[:, :comp]
  projX = np.dot(standardized_df, W)
  reconstructed_pca = np.dot(projX, W.T)
  reconstructed_pca = scaler.inverse_transform(reconstructed_pca)


  plt.plot(original_df.columns, original_df.loc[country_name], color='black', linewidth='4', linestyle='dotted', label='Original')
  plt.plot(original_df.columns, reconstructed[original_df.index.get_loc(country_name)], label='SVD - Non-Standardized')
  plt.plot(original_df.columns, reconstructed_std[original_df.index.get_loc(country_name)], linestyle='dashed', label='SVD - Standardized')
  plt.plot(original_df.columns, reconstructed_pca[original_df.index.get_loc(country_name)], linestyle='dotted', label='PCA - Standardized')
  plt.xlabel('Year')
  plt.ylabel('PC Value')
  plt.legend()
  plt.xticks(standardized_df.columns, rotation=45, ha='right')
  plt.title(f'{comp} PC/SVD Reconstruction of ' + country_name + ' Time Series')

  #plot residual errors for all 3 reconstructions
  residuals = original_df.loc[country_name] - reconstructed[original_df.index.get_loc(country_name)]
  residuals_std = original_df.loc[country_name] - reconstructed_std[original_df.index.get_loc(country_name)]
  residuals_pca = original_df.loc[country_name] - reconstructed_pca[original_df.index.get_loc(country_name)]
  #print(residuals)
  plt.figure(figsize=(15, 5))
  plt.plot(residuals.index, residuals.values)
  plt.plot(residuals_std.index, residuals_std.values, linestyle='dashed')
  plt.plot(residuals_pca.index, residuals_pca.values, linestyle='dotted')
  plt.xlabel('Year')
  plt.ylabel('Residual Error')
  plt.legend(['SVD: Non-Standardized', 'SVD: Standardized', 'PCA: Standardized'])
  plt.xticks(standardized_df.columns, rotation=45, ha='right')
  plt.title('Residual Error of ' + country_name + ' Time Series for Best Reconstruction ' + str(comp) + ' PCs/SVs')
  plt.show()

  #compare using RMSE
  rmse = []
  rmse_std = []
  rmse_pca = []
  num_components = np.arange(1, 17)
  #print(num_components)
  for i in num_components:
    country_index = original_df.index.get_loc(country_name)
    #SVD reconstruction - non-standardized
    reconstructed = U[:, :i] @ np.diag(S[:i]) @ V[:i, :]
    rmse.append(np.sqrt(np.mean((original_df.loc[country_name] - reconstructed[country_index])**2)))

    #SVD Reconstruction - standardized
    reconstructed_std = U_std[:, :i] @ np.diag(S_std[:i]) @ V_std[:i, :]
    reconstructed_std = scaler.inverse_transform(reconstructed_std)
    rmse_std.append(np.sqrt(np.mean((original_df.loc[country_name] - reconstructed_std[country_index])**2)))

    #PCA Reconstruction
    W = eigenvecs[:, :i]
    projX = np.dot(standardized_df, W)
    reconstructed_pca = np.dot(projX, W.T)
    reconstructed_pca = scaler.inverse_transform(reconstructed_pca)
    rmse_pca.append(np.sqrt(np.mean((original_df.loc[country_name] - reconstructed_pca[country_index])**2)))

  #plot RMSE as a function of number of components
  plt.figure(figsize=(15, 5))
  plt.plot(num_components, rmse, label = 'SVD - Non-Standardized')
  plt.plot(num_components, rmse_std, label = 'SVD - Standardized')
  plt.plot(num_components, rmse_pca, linestyle='dotted', label = 'PCA - Standardized')
  plt.legend()
  plt.xlabel('Number of Components')
  plt.ylabel('RMSE')
  plt.title('RMSE of Reconstruction vs Number of Components - '+ country_name)
  plt.show()
````

````python
compare_PCA_SVD(data_raw,'Thailand')
compare_PCA_SVD(data_raw,'Togo')
compare_PCA_SVD(data_raw,'Tonga')
compare_PCA_SVD(data_raw, 'Greece')
compare_PCA_SVD(data_raw,'Somalia')
compare_PCA_SVD(data_raw,'Italy')
````


**Standardization or covariance computation required? Results Comparison**
---

Standardization is not strictly required for SVD, as SVD can be applied directly onto any data matrix. As we can see in the plots above, the RMSE of the non-standardized data is generally lower, suggesting that for this specific data, standardization is not required as it has results in a higher error. It is interesting to note that with standardization, the RMSE is the same for SVD and PCA. Both methods with standardized vs non-standardized data in the case of SVD illustrate a decreasing trend for RMSE which approaches 0 as the number of components/singular values increase, as expected (since the more components added get closer to a 'perfect' reconstruction). Covariance computation is not required for SVD as it works directly on the data matrix (dataframe we pass in) to decompose the original data matrix X into 3 other matrices (U, S and V) which are then used for reconstruction and capture the variance structure of the data directly. However, the covariance matrix can be derived from the SVD matrices and used for reconstruction where  Covariance = (V * S^2 * V^T) / (n-1). From the results above, we see with non-standardized data on the best reconstruction (16 PCs/singular values) - it always aligns very closely to the original trendline across all countries plotted. In contrast, using PCA/standardized SVD for countries like Togo, Tonga, and Somalia has much more deviation, showing a its not as effective in reconstructing the data. This might be due to the scaling effects introduced by standardization or the different ways in which variance is captured by PCA and SVD. We can also see less residual error using the non-standardized data for most of the countries (besides Greece/Italy) but the magnitude is still significant at certain years, indicating that it may not be effectively grasping the trends at each year.
